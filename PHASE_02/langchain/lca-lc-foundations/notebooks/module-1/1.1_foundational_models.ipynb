## Initialising and invoking a model

```python
# as we're working with openai here, this is the way we make our model
# but incase we're working with another model provider, there could be a little different way
# check Model Provider in the cells below

from langchain.chat_models import init_chat_model

model = init_chat_model(model="gpt-5-nano")
```

```python
# this is simply talking directly to the model
# User --> Model --> Response
# it doesn't use any tools, so its not agentic right now
# we can use this approach for simple tasks like Summarization, Q/A, translation, etc.

response = model.invoke("What's the capital of the Moon?")

response
```

```python
print(response.content)
```

```python
from pprint import pprint

pprint(response.response_metadata)
```

## Customising your Model

```python
model = init_chat_model(
    model="gpt-5-nano",
    # Kwargs passed to the model:
    temperature=1.0
)

response = model.invoke("What's the capital of the Moon?")
print(response.content)
```

## Initialising and invoking an agent

```python
# here we're actually creating an agent that can use its tools (when passed)
# User --> Agent --> Model --> Tools --> Back to Model with available tools 
# --> Model decides which tool to use --> Tool runs --> Result comes to the model --> Model responds

from langchain.agents import create_agent

agent = create_agent(model=model)
```

```python
agent = create_agent(model="claude-sonnet-4-6")
```

```python
agent = create_agent("gpt-5-nano")
```

```python
# this gives us one final result after the response is fully generated

from langchain.messages import HumanMessage

response = agent.invoke(
    {
        "messages": [
            HumanMessage(content="What's the capital of the Moon?")
            ]
    }
)
```

```python
from pprint import pprint

pprint(response)
```

```python
print(response['messages'][-1].content)
```

```python
# this is just for demonstration as the model is stateless, 
# it doesn't have any memory unless we pass the conversation so that it has context to answer accordingly
# practically we won't be passing HumanMessage and AIMessage like this
# instead we would be storing it in a chat history object or maybe working with LangGraph for memory

from langchain.messages import AIMessage

response = agent.invoke(
    {
        "messages": [
            HumanMessage(content="What's the capital of the Moon?"),
            AIMessage(content="The capital of the Moon is Luna City."),
            HumanMessage(content="Interesting, tell me more about Luna City")
            ]
    }
)

pprint(response)
```

## Streaming Output

```python
# the reason we're saying 'for token, metadata in agent.stream()' is that 
# the agent.stream() itself yields the AIMessageChunk object that contains the token and the metadata inside it.
# metadata = where that token/chunk came from (tools, model, other node)
# so with this for loop we're actually unpacking the AIMessageChunk with this for loop syntax.
# if we just do 'for token in agent.stream()' it would yield the AIMessageChunk(token and the metadata) in one iteration
# and then we need to manually unpack them. to avoid that, we use the 'for token, metadata in agent.stream() syntax

for token, metadata in agent.stream(
    {"messages": [HumanMessage(content="Tell me all about Luna City, the capital of the Moon")]},
    stream_mode="messages"
):

    # token is a message chunk with token content. It can be 'Moon', ' moon' or 'Luna is a city'. 
    # Token is not always one english word
    # metadata contains which node produced the token
    
    if token.content:  # Check if there's actual content cuz token.content might be empty or None sometimes
        print(token.content, end="", flush=True)  # Print token
```
